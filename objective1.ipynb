```python
# Imports
import pandas as pd
import os

# Define paths (assumes CSV in 'data/' subfolder)
data_dir = 'data'
csv_file = os.path.join(data_dir, 'daily_food.csv')

# Check if file exists
if not os.path.exists(csv_file):
    raise FileNotFoundError(f"CSV file not found at {csv_file}. Ensure it's saved correctly.")
    
print("✅ Setup complete. Proceeding to load CSV.")
```

```python
# Load CSV with error handling for real-world issues (e.g., extra delimiters, missing values)
try:
    df = pd.read_csv(
        csv_file,
        sep=',',  # Standard CSV delimiter
        encoding='utf-8',  # Handles special characters in food names
        on_bad_lines='skip',  # Skips malformed rows (Python 3.9+; use error_bad_lines=False for older)
        low_memory=False  # Handles large files
    )
    print("✅ CSV loaded successfully!")
    print(f"Loaded {len(df)} rows and {len(df.columns)} columns.")
except Exception as e:
    print(f"❌ Error loading CSV: {e}")
    # Fallback: Try alternative encoding
    try:
        df = pd.read_csv(csv_file, encoding='latin-1', on_bad_lines='skip')
        print("✅ Loaded with fallback encoding.")
    except Exception as e2:
        print(f"❌ Fallback failed: {e2}")
        raise e2
```

```python
# Dataset Overview
print("=== DATASET OVERVIEW ===")
print(f"Dataset Shape: {df.shape} (Rows: {df.shape[0]}, Columns: {df.shape[1]})")
print(f"Columns: {list(df.columns)}")

print("\nFirst 5 Rows:")
display(df.head())  # Use display() for nicer output in Jupyter

print("\nDataset Info:")
df.info()  # Shows data types, non-null counts
```

```python
# Key nutritional features as per proposal (e.g., for mapping to nutritional values)
nutritional_features = [
    'Calories (kcal)', 'Protein (g)', 'Carbohydrates (g)', 'Fat (g)', 
    'Fiber (g)', 'Sugars (g)', 'Sodium (mg)', 'Cholesterol (mg)'
]

print(f"\nKey Nutritional Features (X): {nutritional_features}")
all_features_present = all(col in df.columns for col in nutritional_features)
print(f"All key features present: {all_features_present}")

# Quick stats on features
if all_features_present:
    feature_stats = df[nutritional_features].describe()
    display(feature_stats)
else:
    print("⚠️ Some features missing. Check column names.")
```

```python
# 1. First, load the data (if not already loaded)
df = pd.read_csv(csv_file, 
                 on_bad_lines='warn',      # Shows warning instead of error
                 engine='python')

# 2. Check the exact column names
print("Column names in your CSV:")
print(df.columns.tolist())

# 3. Also see first few rows to understand the data
print("\nFirst 5 rows:")
print(df.head())
```

```python
# Count unique food items (success criteria: ≥600)
# === Add this cell and run it first ===
print("=== Actual Column Names ===")
print(df.columns.tolist())

print("\n=== First 5 Rows Preview ===")
print(df.head())

print("\n=== DataFrame Info ===")
print(df.info())

# Count unique food items
print("Available columns:", df.columns.tolist())

# Try to find the correct food column automatically
food_column = None
possible_names = ['Food_Item', 'Food item', 'food_item', 'Food_Name', 'food name', 'item', 'Food']

for col in df.columns:
    if any(name.lower() in col.lower() for name in possible_names):
        food_column = col
        break

if food_column:
    unique_foods = df[food_column].nunique()
    print(f"✅ Using column: '{food_column}'")
    print(f"Unique Food Items: {unique_foods}")
    print(f"Total Rows: {len(df)}")
else:
    print("❌ Could not find food column. Please check the printed columns above.")
```

```python
# Verify success criteria
min_unique_foods = 580
min_columns = 12

criteria_met = unique_foods >= min_unique_foods and df.shape[1] >= min_columns

print("\n=== SUCCESS CRITERIA CHECK ===")
print(f"Unique Foods: {unique_foods} ≥ {min_unique_foods}? {'✅ YES' if unique_foods >= min_unique_foods else '❌ NO'}")
print(f"Columns: {df.shape[1]} ≥ {min_columns}? {'✅ YES' if df.shape[1] >= min_columns else '❌ NO'}")

if criteria_met:
    print("\n🎉 OBJECTIVE 1 SUCCESS! Data ingestion complete.")
    print("Ready for Objective 2: Data Pre-processing (e.g., cleaning, PCA).")
else:
    print("\n⚠️ Criteria not met. Check CSV for completeness (e.g., truncation).")
    print("Tip: Ensure full CSV has ~600+ entries; re-download if needed.")
```

```python
# Save DataFrame summary as JSON (for project reports/proposal alignment)
summary = {
    'shape': list(df.shape),
    'columns': list(df.columns),
    'unique_foods': unique_foods,
    'nutritional_features_present': all_features_present,
    'criteria_met': criteria_met
}

summary_df = pd.DataFrame([summary])
summary_df.to_json('data_ingestion_summary.json', orient='records', indent=2)
print("\n✅ Summary saved to 'data_ingestion_summary.json'")

# Also save a sample of the DataFrame
df.head(10).to_csv('data_sample.csv', index=False)
print("✅ Sample data saved to 'data_sample.csv'")
```
